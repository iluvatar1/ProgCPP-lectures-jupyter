## Matrix exponential from taylor expansion

In scalar calculus, the exponential function $e^x$ can be written as an infinite sum using its Taylor series (specifically, the Maclaurin series) centered at $0$:

$$e^x = \sum_{k=0}^{\infty} \frac{x^k}{k!} = 1 + x + \frac{x^2}{2!} + \frac{x^3}{3!} + \dots$$

We can extend this exact concept to linear algebra. If we have a square matrix $A$ (of size $n \times n$), the **matrix exponential**, denoted as $e^A$, is defined by substituting the matrix into the same Taylor expansion.

---

### The Definition

Since we are dealing with matrices, the scalar $1$ becomes the identity matrix $I$:

$$e^A = \sum_{k=0}^{\infty} \frac{A^k}{k!} = I + A + \frac{A^2}{2!} + \frac{A^3}{3!} + \dots$$

Where:
* $A^0 = I$ (Identity matrix)
* $A^2 = A \times A$
* $k!$ is the factorial of $k$ ($k \times (k-1) \times \dots \times 1$)

```{tip} Why is this important? 
Matrix exponentials are crucial for solving systems of linear differential equations. For example, the system $\frac{d\mathbf{x}}{dt} = A\mathbf{x}$ has the elegant solution $\mathbf{x}(t) = e^{At}\mathbf{x}(0)$.
```

---

### How to Compute It (An Example)

While the infinite sum is the definition, computing it directly by adding infinite matrices is usually impossible unless the matrix has special properties. 

#### Scenario 1: Nilpotent Matrices (The Series Terminates)
If a matrix becomes the zero matrix when raised to a certain power (called a nilpotent matrix), the Taylor series becomes finite and easy to calculate.

Let $A = \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}$. 
Let's find $A^2$:
$$A^2 = \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix} \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix} = \begin{pmatrix} 0 & 0 \\ 0 & 0 \end{pmatrix}$$

Because $A^2 = 0$, all higher powers ($A^3, A^4, \dots$) are also $0$. The infinite Taylor series collapses into just the first two terms:

$$e^A = I + A = \begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix} + \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix} = \begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix}$$

#### Scenario 2: Diagonal Matrices
If a matrix is diagonal, you can simply take the exponential of each diagonal element:

$$\text{If } D = \begin{pmatrix} a & 0 \\ 0 & b \end{pmatrix}, \text{ then } e^D = \begin{pmatrix} e^a & 0 \\ 0 & e^b \end{pmatrix}$$

For more complex matrices, mathematicians usually diagonalize the matrix ($A = PDP^{-1}$) so that $e^A = Pe^DP^{-1}$.

---

### Key Properties to Keep in Mind

* **Non-Commutativity:** In scalar math, $e^x e^y = e^{x+y}$. In matrix math, this is **only true if the matrices commute** ($AB = BA$). If they do not commute, $e^A e^B \neq e^{A+B}$.
* **Inverse:** The inverse of $e^A$ is always $e^{-A}$.
* **Determinant:** $\det(e^A) = e^{\text{trace}(A)}$.

### Typical numerical methods to solve the eigen problem


### Power method
* **Idea:** Multiplying by $A$ several times
* **Why it works:** If $A$ has eigenvalues $\lambda_i$, and they are organized by magnitude, the largest one will dominate.
* **Use:** Finds the eigenvalue with the **largest magnitude** (and its eigenvector).

#### Inverse Iteration
* **Idea:** Instead of multiplying by $A$, we multiply by $A^{-1}$.
* **Why it works:** If $A$ has eigenvalues $\lambda_i$, then $A^{-1}$ has eigenvalues $\frac{1}{\lambda_i}$. The smallest eigenvalue of $A$ becomes the dominant eigenvalue of $A^{-1}$.
* **Use:** Finds the eigenvalue with the **smallest magnitude** (and its eigenvector).

#### Shifted Inverse Iteration & Rayleigh Quotient Iteration
* **Idea:** Apply inverse iteration to $(A - \mu I)^{-1}$, where $\mu$ is a scalar shift.
* **Why it works:** The eigenvalues of $(A - \mu I)^{-1}$ are $\frac{1}{\lambda_i - \mu}$. The closer $\mu$ is to an eigenvalue $\lambda_k$, the larger $\frac{1}{\lambda_k - \mu}$ becomes.
* **Rayleigh Quotient Iteration:** Updates $\mu$ at each step using the current estimate of the eigenvalue: $\mu_k = \frac{x_k^T A x_k}{x_k^T x_k}$. It achieves **cubic convergence** (extremely fast).

#### The QR Algorithm (The Workhorse for Dense Matrices)
* **Idea:** Used by standard libraries (like LAPACK, NumPy's `np.linalg.eig`, MATLAB) to compute **all** eigenvalues and eigenvectors of a full matrix.
* **How it works:**
  1. Decompose $A_k = Q_k R_k$ using the QR decomposition (where $Q$ is orthogonal, $R$ is upper triangular).
  2. Multiply them in reverse order: $A_{k+1} = R_k Q_k$.
* Notice that $A_{k+1} = Q_k^T A_k Q_k$, meaning every step is a similarity transform preserving eigenvalues.
* As $k \to \infty$, $A_k$ converges to an upper triangular matrix (the Schur form). The eigenvalues simply appear on the main diagonal.

#### Lanczos and Arnoldi Algorithms (Krylov Subspace Methods)
* **Idea:** Designed for **very large, sparse matrices** where finding all eigenvalues is too expensive, and you only want a few of the extreme ones (e.g., top $k$ principal components in big data).
* **Arnoldi iteration:** Builds an orthonormal basis for the Krylov subspace $\mathcal{K}_k(A, b) = \text{span}\{b, Ab, A^2b, \dots, A^{k-1}b\}$, projecting the large matrix into a small upper Hessenberg matrix.
* **Lanczos iteration:** A simplified, highly efficient version of Arnoldi specifically for **symmetric matrices** (like the covariance matrix in PCA), reducing the matrix to a small tridiagonal matrix whose eigenvalues are trivial to extract.

#### Jacobi Eigenvalue Method
* **Idea:** Specifically for real symmetric matrices.
* **How it works:** Iteratively applies a sequence of 2D planar rotations (Givens rotations) to systematically eliminate off-diagonal elements until the matrix becomes diagonal. It is slower for large matrices than QR, but highly accurate and easily parallelizable.


#### The power method in detail

The power method is designed to find the **dominant** eigenpair—meaning the eigenvalue with the strictly largest magnitude $|\lambda_1|$ and its associated eigenvector $v_1$.

##### How the Power Method Works (The Core Intuition)

Suppose $A$ is an $m \times m$ matrix with eigenvalues sorted by absolute magnitude:
\begin{equation}
|\lambda_1| > |\lambda_2| \ge |\lambda_3| \ge \dots \ge |\lambda_m|
\end{equation}

Let their corresponding eigenvectors be $v_1, v_2, \dots, v_m$. Assuming these form a basis for $\mathbb{R}^m$, we can write **any** arbitrary initial vector $b_0$ as a linear combination of these eigenvectors:
\begin{equation}
b_0 = c_1 v_1 + c_2 v_2 + \dots + c_m v_m
\end{equation}

Now, consider what happens if we repeatedly multiply $b_0$ by the matrix $A$:

1. After 1 multiplication:
   \begin{equation}
   A b_0 = c_1 A v_1 + c_2 A v_2 + \dots + c_m A v_m = c_1 \lambda_1 v_1 + c_2 \lambda_2 v_2 + \dots + c_m \lambda_m v_m
   \end{equation}

2. After $k$ multiplications:
   \begin{equation}
   A^k b_0 = c_1 \lambda_1^k v_1 + c_2 \lambda_2^k v_2 + \dots + c_m \lambda_m^k v_m
   \end{equation}


Now, factor out $\lambda_1^k$:
\begin{equation}
A^k b_0 = \lambda_1^k \left( c_1 v_1 + c_2 \left(\frac{\lambda_2}{\lambda_1}\right)^k v_2 + \dots + c_m \left(\frac{\lambda_m}{\lambda_1}\right)^k v_m \right)
\end{equation}

Check closely at the ratios $\left(\frac{\lambda_j}{\lambda_1}\right)^k$:
* Since $|\lambda_1| > |\lambda_j|$ for all $j \ge 2$, the ratio $|\lambda_j / \lambda_1| < 1$.
* As $k \to \infty$, every term $\left(\frac{\lambda_j}{\lambda_1}\right)^k \to 0$.

Therefore, for large $k$:
\begin{equation}
A^k b_0 \approx \lambda_1^k c_1 v_1
\end{equation}

Notice that $\lambda_1^k c_1$ is just an overall scalar. This means the vector $A^k b_0$ aligns almost purely in the direction of the dominant **eigenvector** $v_1$!

---

##### The Practical Algorithm (Avoiding Overflow/Underflow)

If you simply compute $A^k b_0$, the vector's magnitude will either explode to $\infty$ (if $|\lambda_1| > 1$) or vanish to $0$ (if $|\lambda_1| < 1$). 

To prevent this numerical instability, we **normalize the vector at each iteration**:

1. **Initialize:** Choose a random unit vector $b_0$ (so $\|b_0\| = 1$).
2. **Iterate:** For $k = 1, 2, 3, \dots$ until convergence:
   $$\tilde{b}_{k} = A b_{k-1} \quad \text{(apply the matrix)}$$
   $$b_k = \frac{\tilde{b}_k}{\|\tilde{b}_k\|} \quad \text{(normalize to unit length)}$$
3. **Extract the Eigenvector:** As $k \to \infty$, $b_k \to v_1$ (the dominant eigenvector).
4. **Extract the Eigenvalue:** Once $b_k \approx v_1$, compute the **Rayleigh quotient**:
   $$\lambda_1 \approx \frac{b_k^T A b_k}{b_k^T b_k} = b_k^T (A b_k) = b_k^T \tilde{b}_k$$

##### Key Notes on Convergence:
* **Convergence rate:** The error drops at a rate of $O\left(\left|\frac{\lambda_2}{\lambda_1}\right|^k\right)$. If $\lambda_1$ is much larger than $\lambda_2$, it converges very quickly. If $|\lambda_1| \approx |\lambda_2|$, it converges slowly.
* **Failure mode:** If the initial guess happens to have $c_1 = 0$ (zero projection on $v_1$), it theoretically fails. In practice, choosing $b_0$ randomly and computer floating-point roundoff errors will inevitably introduce a tiny non-zero component along $v_1$, which will eventually dominate.

### Solving a general eigen value problem with eigen c++
Following the documentation of the general eigen solver,
https://eigen.tuxfamily.org/dox/classEigen_1_1EigenSolver.html, we can use the
example to show how it works:

```c++
#include <iostream>
#include <string>
#include <Eigen/Dense>
#include <complex>

int main(int argc, char **argv) {

    // read the matrix ncols, to be NxN
    const int N = std::stoi(argv[1]);

    // create a random matrix
    // seed controlled by srand
    Eigen::MatrixXd A = Eigen::MatrixXd::Random(N,N);
    std::cout << "Here is a random matrix, A:" << std::endl 
            << A << std::endl << std::endl;

    // create an eigen solver with that matrix. 
    // Once created it solves everything inmediatly
    Eigen::EigenSolver<Eigen::MatrixXd> es(A);

    // print some results
    std::cout << "The eigenvalues of A are:" << std::endl 
            << es.eigenvalues() << std::endl;
    std::cout << "The matrix of eigenvectors, V, is:" << std::endl 
            << es.eigenvectors() << std::endl << std::endl;

    // Check the first eigen value: Av = lambda v
    std::complex<double> lambda = es.eigenvalues()[0];
    std::cout << "Consider the first eigenvalue, lambda = " << lambda << std::endl;
    Eigen::VectorXcd v = es.eigenvectors().col(0);
    std::cout << "If v is the corresponding eigenvector, then lambda * v = " << std::endl 
            << lambda * v << std::endl;
    std::cout << "... and A * v = " << std::endl 
            << A.cast<std::complex<double> >() * v << std::endl << std::endl;

    // diagnalization check
    Eigen::MatrixXcd D = es.eigenvalues().asDiagonal();
    Eigen::MatrixXcd V = es.eigenvectors();
    std::cout << "Finally, V * D * V^(-1) = " << std::endl 
            << V * D * V.inverse() << std::endl;
    
    return 0;
}
```

    std::cout << V * D * V.inverse() << "\n";

    return 0;
}

## Normal modes and matrices
In physics and engineering, **normal modes** describe the natural, synchronized patterns of vibration in an interconnected system (like coupled pendulums, buildings during an earthquake, or atoms in a molecule). 

Linear algebra—specifically **matrices, eigenvalues, and eigenvectors**—is the mathematical engine used to find these modes.

---

### What is a Normal Mode?
In a normal mode, **all parts of the system vibrate at the exact same frequency and pass through their equilibrium positions at the exact same time.** If you have a system with $n$ degrees of freedom (e.g., $n$ masses connected by springs), the system will have exactly $n$ distinct normal modes. Any complex, chaotic-looking motion of the system is actually just a combination (superposition) of these fundamental normal modes.

---

### How Matrices Represent the System

Consider two identical masses, $m$, connected to each other and to walls by identical springs with spring constant $k$. 

If we apply Newton's second law ($F = ma$) to both masses, we get a system of coupled differential equations:

$$m \frac{d^2x_1}{dt^2} = -kx_1 - k(x_1 - x_2) = -2kx_1 + kx_2$$
$$m \frac{d^2x_2}{dt^2} = -kx_2 - k(x_2 - x_1) = kx_1 - 2kx_2$$

We can rewrite this entire system beautifully using a single matrix equation:

$$\begin{pmatrix} m & 0 \\ 0 & m \end{pmatrix} \begin{pmatrix} \ddot{x}_1 \\ \ddot{x}_2 \end{pmatrix} + \begin{pmatrix} 2k & -k \\ -k & 2k \end{pmatrix} \begin{pmatrix} x_1 \\ x_2 \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$

More compactly, this is written as:
$$M\ddot{\mathbf{x}} + K\mathbf{x} = 0$$

Where:
* $M$ is the **Mass Matrix**
* $K$ is the **Stiffness Matrix** (notice how the off-diagonal $-k$ terms "couple" the motion of the two masses together).

---

### Solving for Modes using Eigenvalues

To find a normal mode, we look for a solution where all parts vibrate sinusoidally at the same frequency $\omega$:
$$\mathbf{x}(t) = \mathbf{v} \cos(\omega t)$$

If we take the second derivative of this guess ($\ddot{\mathbf{x}} = -\omega^2 \mathbf{v} \cos(\omega t)$) and plug it back into our matrix equation, the calculus vanishes, leaving us with a pure linear algebra problem:

$$K\mathbf{v} = \omega^2 M\mathbf{v}$$

If we assume $M$ is the identity matrix (or multiply by $M^{-1}$), this becomes a classic **Eigenvalue Problem**:

$$A\mathbf{v} = \lambda\mathbf{v}$$

Where:
* The matrix $A = M^{-1}K$
* The eigenvalue $\lambda = \omega^2$ gives the **frequency** of the vibration ($\omega = \sqrt{\lambda}$).
* The eigenvector $\mathbf{v}$ gives the **shape** of the vibration (the relative displacement of each mass).

---

### The Physical Meaning of the Eigenvectors

For the two-mass system described above, solving the matrix equation yields two eigenvalues and two corresponding eigenvectors:

#### Mode 1: Symmetric Motion (Low Frequency)
* **Eigenvalue:** $\lambda_1 = \frac{k}{m} \implies \omega_1 = \sqrt{\frac{k}{m}}$
* **Eigenvector:** $\mathbf{v}_1 = \begin{pmatrix} 1 \\ 1 \end{pmatrix}$
* **Physical Meaning:** The masses move in the exact same direction with the exact same amplitude. Because they move together, the middle spring never stretches or compresses, meaning the system vibrates at a lower frequency.

#### Mode 2: Antisymmetric Motion (High Frequency)
* **Eigenvalue:** $\lambda_2 = \frac{3k}{m} \implies \omega_2 = \sqrt{\frac{3k}{m}}$
* **Eigenvector:** $\mathbf{v}_2 = \begin{pmatrix} 1 \\ -1 \end{pmatrix}$
* **Physical Meaning:** The masses move in opposite directions (mirror images). This fiercely compresses and stretches the center spring, resulting in a much higher force and a higher frequency of vibration.

---

### Connecting Back to the Matrix Exponential

Remember how the matrix exponential $e^{At}$ solves $\frac{d\mathbf{x}}{dt} = A\mathbf{x}$? 

For second-order system equations like $M\ddot{\mathbf{x}} + K\mathbf{x} = 0$, we can use Euler's formula extensions with matrices ($\cos(\sqrt{M^{-1}K}t)$) or convert the system into a larger first-order system. Ultimately, the **matrix exponential** provides the analytical trajectory of the system over time, while the **eigenvectors** map out the geometric pathways (the normal modes) that the system naturally flows through.

## Other numerical algebra libraries
Here is a curated list of the most widely recommended numerical libraries in **C++** and **Python**, with links, key features, and how each compares to **Eigen**.

---

### Part 1: Top C++ Numerical Libraries

#### 1. [Armadillo](https://arma.sourceforge.net/)
* **What it is:** A high-level C++ linear algebra library with an API deliberately designed to look and feel like MATLAB.
* **Why recommended:** Extremely easy to read and write. If you are prototyping algorithms or translating MATLAB code to C++, Armadillo has the gentlest learning curve.
* **Comparison with Eigen:**
  * **Syntax:** Armadillo’s syntax is more intuitive for scientific programmers than Eigen’s template-heavy API.
  * **Under the hood:** Eigen implements its own vectorized matrix kernels (and is header-only without mandatory external dependencies). Armadillo acts largely as a high-level wrapper over external backend BLAS/LAPACK (like OpenBLAS, Intel MKL).
  * **When to pick it over Eigen:** When you want simpler code and you already have an optimized BLAS/LAPACK installed on your system.

---

#### 2. [Spectra](https://spectralib.org/)
* **What it is:** A header-only C++ library specifically designed for **large-scale eigenvalue and SVD problems** (a modern C++ redesign of ARPACK).
* **Why recommended:** It computes a few extreme eigenvalues/eigenvectors of large, sparse, or even matrix-free operators (using Krylov subspace methods like Arnoldi and Lanczos).
* **Comparison with Eigen:**
  * **Spectra is built directly on top of Eigen!**
  * Eigen’s built-in `EigenSolver` (which you used earlier) is designed for **dense** matrices and computes *all* $N$ eigenvalues ($O(N^3)$ complexity). If you have a $50,000 \times 50,000$ sparse matrix and only need the top 5 eigenvalues (e.g., PCA on high-dimensional data), Eigen alone will run out of memory or take hours.
  * **When to pick it:** Whenever you need partial eigensolves on large or sparse matrices alongside Eigen.

---

#### 3. [Trilinos](https://trilinos.github.io/)
* **What it is:** A massive collection of open-source packages developed by Sandia National Laboratories for extreme-scale scientific computing and multi-physics simulations.
* **Why recommended:** It is the industry/academic standard for massively parallel, distributed-memory (MPI) engineering simulations (finite elements, nonlinear solvers, PDE solving).
* **Comparison with Eigen:**
  * **Scale:** Eigen is designed for single-machine, shared-memory computations (desktop/server level). Trilinos is designed to run across thousands of nodes on supercomputers.
  * **Complexity:** Trilinos has a steep learning curve and a heavyweight build system (CMake with dozens of interdependent packages like Epetra, Tpetra, Anasazi).
  * **When to pick it over Eigen:** When your problem is so large it must be distributed across an MPI cluster.

---

#### 4. [PETSc](https://petsc.org/) & [SLEPc](https://slepc.upv.es/)
* **What it is:** Developed by Argonne National Laboratory, PETSc (Portable, Extensible Toolkit for Scientific Computation) and its eigenvalue extension SLEPc are the gold standards for scalable PDE and sparse eigenvalue solvers.
* **Why recommended:** Unmatched reliability for massive sparse linear systems, iterative solvers, and nonlinear systems on HPC systems.
* **Comparison with Eigen:**
  * Written in C (with clean C++ bindings). Like Trilinos, it operates at the supercomputer / MPI cluster scale.
  * Eigen is much easier to integrate into small/medium standalone C++ applications (header-only, zero dependencies), whereas PETSc is an entire runtime environment.

---

#### 5. [Blaze](https://bitbucket.org/blaze-lib/blaze)
* **What it is:** A high-performance, open-source C++ math library using smart expression templates.
* **Why recommended:** Regularly benchmarks as one of the fastest dense and sparse linear algebra libraries on modern multi-core x86 architectures, with aggressive SIMD vectorization and OpenMP/threads integration.
* **Comparison with Eigen:**
  * Similar philosophy to Eigen (header-only, expression templates to avoid temporaries).
  * Often slightly outperforms Eigen on specific dense arithmetic benchmarks, but Eigen has a much larger ecosystem, better geometric transformations, and wider geometry/robotics adoption.

---

#### 6. [ArrayFire](https://arrayfire.com/)
* **What it is:** A high-performance GPU and multi-core library with a unified API that runs on CUDA, OpenCL, and CPU.
* **Why recommended:** You write code once, and it automatically accelerates on NVIDIA GPUs, AMD GPUs, or multi-threaded CPUs without rewriting kernel code.
* **Comparison with Eigen:**
  * Eigen is primarily CPU-focused (its CUDA support is limited/experimental). ArrayFire is built from the ground up for GPU acceleration.

---

### Part 2: Top Python Numerical Libraries

| Library | Link | Best Use Case | Relation to Eigen / C++ |
| :--- | :--- | :--- | :--- |
| **NumPy** | [numpy.org](https://numpy.org/) | Foundation for all array/matrix math in Python. | High-level interface wrapping C and Fortran BLAS/LAPACK. |
| **SciPy** | [scipy.org](https://scipy.org/) | Scientific algorithms: sparse matrices, optimization, signal processing. | `scipy.sparse.linalg.eigs` wraps ARPACK (the Python equivalent of what Spectra provides for Eigen). |

| **CuPy** | [cupy.dev](https://cupy.dev/) | Drop-in replacement for NumPy/SciPy targeting NVIDIA CUDA. | Identical API to NumPy, but runs everything on GPUs. |

### Summary: Which Should You Choose?

* **Use Eigen** if: You need a self-contained, header-only C++ library for dense linear algebra, robotics, computer vision, or small-to-medium matrices on a single machine.
* **Add Spectra to Eigen** if: You have large or sparse matrices and only need the top/bottom $k$ eigenvalues (like PCA on large datasets).
* **Use Armadillo** if: You are porting code from MATLAB to C++ and want the simplest possible C++ syntax.
* **Use PETSc / Trilinos** if: Your problem requires an MPI cluster or supercomputer.
* **Use PyTorch / ArrayFire** if: You want to offload matrix operations directly to GPUs with minimal friction.
